# Max-Cut26: shared LFSR / EPIX
Run all cells on PYNQ-Z1 using its hardware-enabled Python kernel. Set `trials` below to 1–1000. Only one notebook may control the FPGA at a time.


```python
# Keep this notebook in its demo folder so the runner finds the matching files.
from pathlib import Path
import json, subprocess, sys, uuid
project = Path.cwd()
assert (project / "run.py").is_file()
trials = 1  # Increase to 1000 for a full campaign.
output = project / "runs" / uuid.uuid4().hex
```

```python
# Use this kernel's Python to load the FPGA and run the selected policy.
subprocess.check_call([sys.executable, str(project / "run.py"), "--mode", "epix", "--trials", str(trials), "--output", str(output)])
```

```python
# successes counts trials that reached the optimum; mean_rng_words counts fresh draws.
summary = json.loads((output / "SUMMARY.json").read_text())
print(json.dumps(summary, indent=2))
print("Saved every trial to", output)
```

Each trial runs all 1024 sweeps. IID draws a fresh LFSR word for every update; EPIX reuses cached words when its reuse conditions are met.
